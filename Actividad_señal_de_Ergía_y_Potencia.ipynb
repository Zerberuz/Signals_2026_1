```python
import numpy as np
import matplotlib.pyplot as plt

#Vector de tiempo
t = np.linspace(-10,10,4000)

#Señal
x = np.sin(3*t)

#Potencia Promedio
E = np.trapz(x**2,t)/(t[-1]-t[0])
print("Potencia promedio aproximada",  E)

#Graficar
plt.plot(t, x, label = 'x(t)=sin(3t)')
plt.title('Señal de Potencia')
plt.xlabel('Tiempo')
plt.ylabel('Amplitud')
plt.grid(True)
plt.legend()
plt.show()
```

#Con el desarrollo del código se comprueba que la señal 𝑥(𝑡)=sin⁡(3𝑡) es una señal de potencia, ya que presenta una potencia promedio finita. El valor obtenido es cercano a 0,5, lo cual coincide con el resultado teórico para señales sinusoidales. Esto confirma que la frecuencia no afecta la potencia promedio, sino únicamente el comportamiento oscilatorio de la señal en el tiempo.

```python
import numpy as np
import matplotlib.pyplot as plt

# Vector de tiempo
t = np.linspace(-5, 5, 400)

# Señal a tramos: x(t) = 1 si |t| <= 1, 0 en otro caso
x = np.piecewise(t,[np.abs(t) <= 1, np.abs(t) > 1],[1, 0])

# Energía aproximada
E = np.trapz(x**2, t)
print("Energía aproximada:", E)

# Gráfica
plt.plot(t, x, label='x(t) = 1 si |t|≤1, 0 en otro caso')
plt.title('Pulso rectangular (Señal de energía)')
plt.xlabel('Tiempo')
plt.ylabel('Amplitud')
plt.grid(True)
plt.legend()
plt.show()
```

# A partir del código se verifica que el pulso rectangular definido por x(t)=0 en otro caso es una señal de energía, ya que su energía es finita. El valor obtenido es cercano a 2, lo cual coincide con el resultado teórico. Esto confirma que el método de integración empleado representa adecuadamente el comportamiento energético de la señal.

```python
import numpy as np
import matplotlib.pyplot as plt

# Vector de tiempo
t = np.linspace(-5, 5, 400)

# Señal a tramos: x(t) = 1 si |t| <= 1, 0 en otro caso
x[(t >= -2) & (t <= 2)] = np.sin(3 * t[(t >= -2) & (t <= 2)])

# Energía aproximada
E = np.trapz(x**2, t)
print("Energía aproximada:", E)

# Gráfica
plt.plot(t, x, label='x(t) = seno(3t) si |t|≤2, 0 en otro caso')
plt.title('Pulso rectangular (Señal de energía)')
plt.xlabel('Tiempo')
plt.ylabel('Amplitud')
plt.grid(True)
plt.legend()
plt.show()
```

# La señal senoidal truncada 𝑥(𝑡)=sin(3𝑡) para ∣𝑡∣≤2 es una señal de energía, debido a que solo existe en un intervalo de tiempo finito. Al estar limitada su energía total es finita y su potencia promedio es cero.
